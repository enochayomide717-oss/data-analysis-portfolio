```python
import pandas as pd
df=pd.read_excel("AR Data.xlsx")
df
df["Month"]=df["Order Date"].apply(lambda x : x.strftime('%B'))
df
df["Year"]=df["Order Date"].apply(lambda x : x.strftime('%Y'))
df
```

```python
#barchart
barchart= df.pivot_table(index="Products",
                         columns="Year",
                         values="Profit",
                         aggfunc="sum")
barchart.plot(kind="barh")
```

```python
#column chart
barchart= df.pivot_table(index="Products",
                         columns="Year",
                         values="Profit",
                         aggfunc="sum")
barchart.plot(kind="bar")
```

```python
#column chart of profits by products with labelling
from matplotlib import pyplot as plt
barchart= df.pivot_table(index="Products",
                         values="Profit",
                         aggfunc="sum")
barchart.plot(kind="bar", color="green")
plt.title("Profit by products")
plt.xlabel("Profits")
plt.ylabel("Products")
```

```python
#column chart of profits by products with labelling
from matplotlib import pyplot as plt
barchart= df.pivot_table(index="Products",
                         values="Profit",
                         aggfunc="sum")
barchart.plot(kind="barh", color="green")
plt.title("Profit by products")
plt.xlabel("Profits")
plt.ylabel("Products")
```

```python
#pie chart of profits by products with labelling
from matplotlib import pyplot as plt
piechart= df.pivot_table(index="Countries",
                         values="Revenue",
                         aggfunc="sum")
piechart.plot(kind="pie",y ='Revenue',autopct='%1.0f%%')
plt.title("Revenue by countries")
```

```python
#sorted column chart
from matplotlib import pyplot as plt
dfpivot=df.pivot_table(index="Month",
                       columns="Year",
                       values="Profit",
                       aggfunc="sum")
month_dict={'January':1,'February':2,'March':3,'April':4,'May':5,'June':6,'July':7,'August':8,'September':9,'October':10,'November':11,'December':12}
dftable=dfpivot.sort_values('Month',key=lambda x :x .apply(lambda x : month_dict[x]))
dftable.plot(kind="bar")

plt.title("Profit by month")
plt.xlabel("Months")
plt.ylabel("Profits")
```
