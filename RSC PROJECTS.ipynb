```python
import pandas as pd
df=pd.read_excel("RSC.xlsx")
df
del df['Status']
df
del df['unnamed1']
df
df["Gender"]=df['Gender'].str.replace('M','Male')
df
df["Gender"]=df['Gender'].str.replace('F','Female')
df
df["Marital_Status"]=df["Marital_Status"].astype(str)
df["Marital_Status"]=df["Marital_Status"].str.replace('1','Married')
df
df["Marital_Status"]=df["Marital_Status"].str.replace('0','Single')
df
dff=df["Amount"]=df["Amount"].fillna(0)
df.head(10)
dff=df
df
```

```python
#sum of orders by occupation and add grand total
dff.pivot_table(index="Occupation",
                values="Orders",
             aggfunc="sum",
                margins=True,
                margins_name='Grand Total')



```

```python
#to sort by occupation(a-z)
t=dff.pivot_table(index="Occupation",
                values="Orders",
             aggfunc="sum")
t.sort_values(by=["Occupation"],ascending=True)
```

```python
t=dff.pivot_table(index="Product_Category",
                values="Amount",
             aggfunc="sum")
Q7=t.sort_values(by=["Amount"],ascending=False).head(10)
Q7["%"]=100*Q7['Amount']/Q7['Amount'].sum()
Q7.round(0)
```
